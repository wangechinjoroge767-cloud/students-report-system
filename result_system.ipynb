```
subject = ["Maths", "English", "Science"] # Global subject list

def get_valid_marks(subject_name):
    while True:
        entry = input(f"Enter marks for {subject_name}: ").strip()
        try:
            mark = float(entry)
        except ValueError:
            print("Invalid input. Please enter a number.")
            continue
        return mark

def get_students_input():
    students = []
    while True:
        name = input(" student name : ").strip()
        if name.lower() == 'done':
            break

        marks = {}
        for s in subject:
            marks[s] = get_valid_marks(s)
        students.append({'name': name, 'marks': marks})
    return students

def get_grade(mark):
    if mark >= 80:
        return "A"
    elif mark >= 65:
        return "B"
    elif mark >= 50:
        return "C"
    elif mark >= 40:
        return "D"
    else:
        return "F"

def calculate_average(student):
    # Ensure marks are floats for calculation, as they might come from input as strings
    total = sum(float(student["marks"][s]) for s in subject)
    return round(total / len(subject), 2)


def get_student_report(student):
    subjects_details = {}
    for s in subject:
        mark = float(student["marks"][s])
        subjects_details[s] = {"mark": mark, "grade": get_grade(mark)}

    average = calculate_average(student)
    return {
        "name": student["name"],
        "subjects": subjects_details,
        "average": average,
        "overall_grade": get_grade(average),
    }


def get_all_results(students):
    return [get_student_report(student) for student in students]


def get_top_student(students):
    if not students:
        return None
    return max(students, key=calculate_average)


def view_all_students(students):
    if not students:
        print("No students added yet.")
        return

    print("\nAll Students:")
    for student in students:
        print(f"Name: {student['name']}, Marks: {student['marks']}")


def print_report(students):
    if not students:
        print("\nNo students to report on.")
        return

    print("\n" + "=" * 68)
    print("                 NAKURU HIGH SCHOOL - RESULTS REPORT")
    print("=" * 68)
    print(f"{'Name':<18} {'Maths':<12} {'English':<12} {'Science':<12} {'Average':<8} Grade")
    print("-" * 68)

    for result in get_all_results(students):
        maths = result["subjects"]["Maths"]
        english = result["subjects"]["English"]
        science = result["subjects"]["Science"]

        maths_text = f"{maths['mark']}({maths['grade']})"
        english_text = f"{english['mark']}({english['grade']})"
        science_text = f"{science['mark']}({science['grade']})"

        print(
            f"{result['name']:<18} "
            f"{maths_text:<12} "
            f"{english_text:<12} "
            f"{science_text:<12} "
            f"{result['average']:<8} "
            f"{result['overall_grade']}"
        )

    top_student = get_top_student(students)
    if top_student:
        top_average = calculate_average(top_student)
        print("-" * 68)
        print(f"Top Student: {top_student['name']} | Average: {top_average}")
    else:
        print("-" * 68)
        print("No top student found.")

    print("Grade Scale: A=80+  B=65+  C=50+  D=40+  F=<40")
    print("=" * 68 + "\n")


def show_menu():
    print("\n" + "=" * 45)
    print("Nakuru High School - Student Results Management System")
    print("=" * 45)
    print("1. Add Student")
    print("2. View All Students")
    print("3. Print Report")
    print("4. Exit")
    print("=" * 45)


def get_menu_choice():
    while True:
        choice = input("Enter your choice (1-4): ").strip()
        if choice in ["1", "2", "3", "4"]:
            return choice
        print("Invalid choice. Please enter a number between 1 and 4.")


def main():
    print("Welcome to Nakuru High School - Student Results Management System")
    students = []

    while True:
        show_menu()
        choice = get_menu_choice()

        if choice == "1":
            students.extend(get_students_input())
        elif choice == "2":
            view_all_students(students)
        elif choice == "3":
            print_report(students)
        elif choice == "4":
            print("Exiting the system. Goodbye!")
            break


if __name__ == "__main__":
    main()
```
